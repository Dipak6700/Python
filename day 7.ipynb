```python
bitString = input("Enter a string of bits")
decimal =0 
power = len(bitString) -1

for digit in bitString:
    decimal = decimal + int(digit) * 2 **power
    power = power -1
print(decimal)


```

```python
decimal = int(input("Enter a decimal number: "))

binary = ""

while decimal > 0:
    remainder = decimal % 2
    binary = str(remainder) + binary
    decimal = decimal // 2

print("Binary number:", binary)
```
